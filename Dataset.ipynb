## 6.数据映射(使用map方法可以添加字段也可以对某一字段中的数据进行添加或者减少)

```python
def add_title(example):
    example["title"] = "title:"+example["title"]
    return example
title_datasets = dataset.map(add_title)
title_datasets["train"][:10]["title"]
```

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-chinese")
def process_function(example,tokenizer = tokenizer):
    model_inputs = tokenizer(example["content"],max_length = 512,truncation = True)
    labels = tokenizer(example["title"],max_length = 32,truncation = True)
    model_inputs["labels"]=labels["input_ids"]
    return model_inputs
processed_dataset = datasets.map(process_function)
processed_dataset
```

```python
# 进行批处理
processed_dataset = datasets.map(process_function,batched = True)
processed_dataset
```

```python
# 当tokenizer无法支持批处理时，可以采用多进程的形式进行快速加载
processed_dataset = datasets.map(process_function,num_proc = True)
processed_dataset
```

```python
# 用map方法实现，删除字段
# remove_columns接收一个字段列表
# column_names这个属性可以直接获得字段列表
processed_dataset = datasets.map(process_function,batched = True,remove_columns=datasets["train"].column_names)
processed_dataset
```

## 10.自定义加载脚本进行加载数据(现版本已经不支持使用自定义的脚本了)

```python
#如果想要加载比较复杂的数据集，就可以使用自定义的加载脚本
dataset = load_dataset("json", data_files="./cmrc2018.json",field = "data")
dataset
```

```python
import json
from datasets import Dataset

with open("./cmrc2018.json", "r", encoding="utf-8") as f:
    raw = json.load(f)

contexts = []
questions = []
ids = []
answer_starts = []
answer_texts = []

for article in raw["data"]:
    for paragraph in article["paragraphs"]:
        context = paragraph["context"]
        for qa in paragraph["qas"]:
            contexts.append(context)
            questions.append(qa["question"])
            ids.append(qa["id"])
            # 提取所有答案的起始位置和文本（列表）
            starts = [ans["answer_start"] for ans in qa["answers"]]
            texts = [ans["text"] for ans in qa["answers"]]
            answer_starts.append(starts)
            answer_texts.append(texts)
#面对列表时使用for循环不断提取文字列表，自己通过读取字典来构建数据集
dataset = Dataset.from_dict({
    "context": contexts,
    "question": questions,
    "id": ids,
    "answer_start": answer_starts,
    "answer_text": answer_texts,
})
dataset
```

```python
from datasets import Dataset, Features, Value, Sequence
# 如果想要自定义数据集的架构的话可以通过features来进行
features = Features({
    "context": Value("string"),
    "question": Value("string"),
    "id": Value("string"),
    "answers": {
        "answer_start": Sequence(Value("int32")),
        "text": Sequence(Value("string")),
    }
})

# 合并两端数据
answers_list = [{"answer_start": starts, "text": texts} for starts, texts in zip(answer_starts, answer_texts)]

# 增加features参数
dataset = Dataset.from_dict({
    "context": contexts,
    "question": questions,
    "id": ids,
    "answers": answers_list,
}, features=features)
dataset
```

## 实战：dataset with datacollator

```python
from transformers import DataCollatorWithPadding

dataset = Dataset.from_csv("./waimai_10k.csv",split="train")
dataset = dataset.filter(lambda x:x["review"] is not None)#对数据集进行筛选
dataset
```

```python
# 数据处理
def process_function(examples):
    tokenized_examples = tokenizer(examples["review"], max_length=128, truncation=True)
    tokenized_examples["labels"] = examples["label"]
    return tokenized_examples
tokenized_dataset = dataset.map(process_function,batched = True , remove_columns = dataset.column_names)#remove_columns = dataset.column_names移除所有的原始列
tokenized_dataset
```

```python
#生成dataloader
collator = DataCollatorWithPadding(tokenizer=tokenizer)
from torch.utils.data import DataLoader
Dl = DataLoader(tokenized_dataset,batch_size = 128,collate_fn = collator,shuffle = True)
next(enumerate(Dl))
```
